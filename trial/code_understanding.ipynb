# Mutabili vs Immutabili

```python
x = "ciao"
y = x
x += " mondo"
print(y)  # Output: ciao. Perchè y punta alla vecchia partizione di memoria di x, quindi non viene modificata quando x cambia.
```

```python
x = (1,2)
y = x
x += (3,)
print(y)  # Output: (1, 2). Perchè y punta alla vecchia partizione di memoria di x, quindi non viene modificata quando x cambia.
```

```python
a = [1,2]
b = a
a.append(3)
print(b)  # Output: [1, 2, 3]. Perchè b punta alla stessa partizione di memoria di a, quindi viene modificata quando a cambia.
```

```python
x = 5
y = x
x += 1
print(y)  # Output: 5. Perchè y punta alla vecchia partizione di memoria di x, quindi non viene modificata quando x cambia.
```

```python
x = 10
def f():
    print(x) # x è interno alla funzione e non è un parametro, quindi darà un errore
    x = 20
f()
```

# Inizializzazione parametri Funzioni

```python
def add_item(item, lst=[]): # il valore di default viene creato una sola volta quando la funzione viene definita, quindi viene condiviso tra le chiamate della funzione.
    lst.append(item)
    print(id(lst))  # Output: <class 'list'>. Perchè lst è una lista mutabile e viene condivisa tra le chiamate della funzione.
    return lst
print(add_item(1))  # Output: [1]
print(add_item(2))  # Output: [1, 2]. Perchè lst è una lista mutabile e viene condivisa tra le chiamate della funzione.


# Come si aggiusta?
def add_item(item, lst=None):
    if lst is None:
        lst = []
    lst.append(item)
    print(id(lst))  # Output: <class 'list'>. Perchè lst è una lista mutabile e viene condivisa tra le chiamate della funzione.
    return lst
print(add_item(1))  # Output: [1]
print(add_item(2))  # Output: [2]. Perchè lst ora è creato internamente alla funzione e non viene condiviso tra le chiamate della funzione.
```

```python
def add_to_list(val, my_list=[]):
    my_list.append(val)
    return my_list

list1 = add_to_list(1)
list2 = add_to_list(123, []) # Qui stiamo generando una nuova lista vuota, quindi non viene condivisa con le altre chiamate della funzione.
list3 = add_to_list('a') # Usa la stessa lista di default creata nella prima chiamata della funzione.
print(list1)
```

Creatore di Funzioni

```python
def create_multipliers():
    return [lambda x: i * x for i in range(4)]  # La variabile i viene catturata dalla chiusura e il suo valore finale è 4.
for m in create_multipliers():
    print(m(2))  # Output: 6, 6, 6, 6. Perchè la variabile i viene catturata dalla chiusura e il suo valore finale è 4.


```
